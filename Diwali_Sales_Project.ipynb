**Exploratory Data Analysis**

---



*Gender*

**Plotting a Bar Chart for Gender and its Count**

```python
ax=sns.countplot(x='Gender',data=a,color='red')
for bars in ax.containers:
  ax.bar_label(bars)
```

# **Plotting a Bar Chart for Gender vs Total_Amount**

```python
ss=a.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False )
sns.barplot(x='Gender',y='Amount',data=ss)
```

**From the above Graphs, We can see that most os the Buyers are females and even the purchesing power of females are greater then men.**

# **Age**

```python
ax=sns.countplot(x='Age Group',data=a,color='grey',hue='Gender')
for bars in ax.containers:
  ax.bar_label(bars)
```

**Total Amount vs Age Group**

```python
sales_age=a.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Age Group',y='Amount',data=sales_age,color='lightpink')
```

**From the above Graph we can observe that most of the buyers are from the age group of 26-35 years.**

# States

**Total Number of Orders from the top 10 States**

```python
sales_age= a.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='State',y='Orders',data=sales_age,color='lightgreen')
```

**From the above graphs we can see that most of the the orders and total sales /Amount are from Uttar Pradesh,Maharashtra, Karnataka respectively.**

# **Marital Status**

```python
ac=sns.countplot(data=a,x='Marital_Status',color='orange')
sns.set(rc={'figure.figsize':(2,5)})
for bars in ac.containers:
  ac.bar_label(bars)

```

```python
sales_age=a.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(5,5)})
sns.barplot(x='Marital_Status',y='Amount',data=sales_age,hue='Gender')
```

**From the above graphs We can see that most of the buyers are married(women) and they have high perchasing power.**

**OCCUPATION**

```python
sns.set (rc={'figure.figsize':(23,5)})
ax=sns.countplot(data=a,x='Occupation',color='orange')
for bars in ax.containers:
  ax.bar_label(bars )
```

```python
Sales_state=a.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Occupation',y='Amount',data=Sales_state,color='lightgreen')
```

**From the above graph we can see that most of the buyers are itsector. healtcare and aviation center..**

**PRODUCT CATEGORY**

```python
sns.set(rc={'figure.figsize':(20,5)})
ax=sns.countplot(data=a,x='Product_Category',color='red')
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
Sales_state=a.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(24,5)})
ax=sns.barplot(x='Product_Category',y='Amount',data=Sales_state,color='salmon')
for bars in ax.containers:
  ax.bar_label(bars)
```

**From the above graphs we can see that most of the sold products are from the food, clothing and electronic categories..**

```python
Sales_s=a.groupby(['Product_ID'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,5)})
ax=sns.barplot(x='Product_ID',y='Orders',data=Sales_s,color='orange')
for bars in ax.containers:
  ax.bar_label(bars )
```

# **Top 10 Most Sold Products**

```python
fig1,ax1=plt.subplots(figsize=(12,7))
a.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False).plot(kind='bar')
```

**Married Women From Age group of 26-35 from UP, Maharashtra ,Karnatka working in IT, Healthcare,Aviation are more liable to buy product from Food, Clothing And Electronic Categories.**
